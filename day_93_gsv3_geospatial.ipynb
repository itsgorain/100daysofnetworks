# Businesses Near One Business
This is old code from 2025 when I was first experimenting with GrooveSeeker data. Let's use it as a baseline and then make something useful.

```python
address = '2724 E Burnside St, Portland, OR 97214' # Rose City Guitar Company
```

```python
location = ox.geocode(address)
print(f"Coordinates: {location}")
```

```python
max_distance = 1000

tags = {
    'amenity': True,  # Gets ALL amenities
    'shop': True,     # Gets ALL shops
    'tourism': True,  # Gets ALL tourism
    'leisure': True   # Gets ALL leisure
}

businesses = ox.features_from_address(address, tags, dist=max_distance)
businesses_filtered = businesses.dropna(subset=['name'])

print("Found {} businesses within {}m".format(len(businesses_filtered), max_distance))

for idx, row in businesses_filtered.iterrows():
    #print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
    #print('------------------------------------')
    print()
    print("NAME: {}".format(row.get('name', 'Unknown')))
    
    # Check for amenity, shop, tourism, or leisure type
    entity_type = row.get('amenity') or row.get('shop') or row.get('tourism') or row.get('leisure')
    if entity_type and str(entity_type) != 'nan':
        print("Type: {}".format(entity_type))
    
    street = row.get('addr:street')
    number = row.get('addr:housenumber')
    if street and str(street) != 'nan' and number and str(number) != 'nan':
        print("Address: {} {}".format(number, street))
    elif street and str(street) != 'nan':
        print("Address: {}".format(street))
    
    cuisine = row.get('cuisine')
    if cuisine and str(cuisine) != 'nan':
        print("Cuisine: {}".format(cuisine))
    
    phone = row.get('phone')
    if phone and str(phone) != 'nan':
        print("Phone: {}".format(phone))
```

```python

# Get address coordinates
address_coords = ox.geocode(address)

# Create interactive map with light basemap
m = folium.Map(
    location=address_coords,
    zoom_start=18,
    tiles='OpenStreetMap'  # Light, readable basemap
)

# Blue marker for the address
folium.CircleMarker(
    address_coords,
    radius=10,
    popup='<b>{}</b>'.format(address),
    color='#2E86DE',
    fill=True,
    fillColor='#2E86DE',
    fillOpacity=1.0,
    weight=3
).add_to(m)

# Red markers for businesses
for idx, row in businesses_filtered.iterrows():
    if 'geometry' in row and row['geometry']:
        coords = [row['geometry'].centroid.y, row['geometry'].centroid.x]
        name = row.get('name', 'Unknown')
        entity_type = row.get('amenity') or row.get('shop') or row.get('tourism') or row.get('leisure')
        
        folium.CircleMarker(
            coords,
            radius=6,
            popup='<b>{}</b><br>{}'.format(name, entity_type),
            color='#EE5A6F',
            fill=True,
            fillColor='#EE5A6F',
            fillOpacity=0.8
        ).add_to(m)

m
```

# Overlay Other Points of Interest

Now, I want to overlay all other known record stores as blue dots so that I can easily find more places to drop off GrooveSeeker event papers.

```python
# Record Stores

import pandas as pd

data = 'data/verdant_eye/gsv3/record_stores_pdx.csv'

record_store_df = pd.read_csv(data)

record_store_df['full_address'] = record_store_df['Address'] + ', ' + record_store_df['City'] + ', ' + record_store_df['State']
record_store_df = record_store_df[['Company Name', 'full_address']]
record_store_df.columns = ['company', 'address']
record_store_df.head()
```

```python
# Music Stores

data = 'data/verdant_eye/gsv3/music_stores_pdx.csv'

music_store_df = pd.read_csv(data)

music_store_df['full_address'] = music_store_df['Address'] + ', ' + music_store_df['City'] + ', ' + music_store_df['State']
music_store_df = music_store_df[['Company Name', 'full_address']]
music_store_df.columns = ['company', 'address']
music_store_df.head()
```

```python
# Venues

data = 'data/verdant_eye/gsv3/venues_pdx.csv'

venue_df = pd.read_csv(data)

venue_df['full_address'] = venue_df['Address'] + ', ' + venue_df['City'] + ', ' + venue_df['State']
venue_df = venue_df[['Company Name', 'full_address']]
venue_df.columns = ['company', 'address']
venue_df.head()
```

```python
import osmnx as ox
import folium

# Geocode all spots
for name, data in spots.items():
    coords = ox.geocode(data['address'])
    spots[name]['coords'] = coords

# Geocode check_df rows
def safe_geocode(address):
    try:
        return ox.geocode(address)
    except Exception:
        return None

record_store_df['coords'] = record_store_df['address'].apply(safe_geocode)
record_store_df = record_store_df[record_store_df['coords'].notna()]

music_store_df['coords'] = music_store_df['address'].apply(safe_geocode)
music_store_df = music_store_df[music_store_df['coords'].notna()]

venue_df['coords'] = venue_df['address'].apply(safe_geocode)
venue_df = venue_df[venue_df['coords'].notna()]

# Build map centered on Portland
m = folium.Map(location=[45.5231, -122.6765], zoom_start=13, tiles='OpenStreetMap')

# blue dots: record stores
for _, row in record_store_df.iterrows():
    folium.CircleMarker(
        row['coords'],
        radius=8,
        popup='<b>{}</b><br>{}'.format(row['company'], row['address']),
        color='#2E86DE',
        fill=True,
        fillColor='#2E86DE',
        fillOpacity=0.8
    ).add_to(m)

# green dots: music stores
for _, row in music_store_df.iterrows():
    folium.CircleMarker(
        row['coords'],
        radius=8,
        popup='<b>{}</b><br>{}'.format(row['company'], row['address']),
        color='#2ECC71',
        fill=True,
        fillColor='#2ECC71',
        fillOpacity=0.8
    ).add_to(m)

# purple dots: venues
for _, row in venue_df.iterrows():
    folium.CircleMarker(
        row['coords'],
        radius=8,
        popup='<b>{}</b><br>{}'.format(row['company'], row['address']),
        color='#9B59B6',
        fill=True,
        fillColor='#9B59B6',
        fillOpacity=0.8
    ).add_to(m)

# red dots: paper drop spots
for name, data in spots.items():
    folium.CircleMarker(
        data['coords'],
        radius=8,
        popup='<b>{}</b><br>{}'.format(name, data['address']),
        color='#EE5A6F',
        fill=True,
        fillColor='#EE5A6F',
        fillOpacity=0.8
    ).add_to(m)

m
```
